```python
# Classes + special methods + Docstring
class TableWithLists:
  """
  A simple table implemented with lists.

  Atributes:
    table(list): a list containing as many elements as the table has rows, which each element a list of as many elements the table has columns.

  """
  def __init__(self,
               n_columns: int,
               n_rows: int,
               default_value: int | float | str = 0) -> None:
    """
    Initialise the table.

    Args:
      n_columns (int): number of columns in the table.
      n_rows (int): number of rows in the table.
      default_value (int | float | str): default value for the elements of the table.

    Return:
      None.
    """
    self.table = []
    for i in range(n_rows):
      row = []
      for j in range(n_columns):
        row.append(default_value)
      self.table.append(row)

  def __str__(self)->str:
    """
    Special methods invoked when the object is called in a print().

    Args:
      None.

    Return:
      table_to_print (str): a string containing the elements of the table, organised by rows and columns.
    """
    table_to_print = ""
    for row in self.table:
      table_to_print += str(row) + "\n"
    return table_to_print


test_table = TableWithLists(3,4)
print(test_table)
```

```python
# Create a table with ''UID','Combined_Key','Date','Deaths','Confirmed' for Utah and Colorado
samples_table = dataset.copy()
samples_table = samples_table[(samples_table['Province_State']=='Utah')|(samples_table['Province_State']=='Colorado')]
samples_table = samples_table[['UID','Combined_Key','Date','Deaths','Confirmed']]
print(samples_table.shape)
```

```python
# Inner Join : contains only those rows that have matching values in both of the original DataFrames
merged_inner = pd.merge(left=samples_table, right=geographic_info_table, left_on='Combined_Key', right_on='Combined_Key')
merged_inner
```

```python
# Left Join : all of the rows from the left DataFrame, even those rows whose join key(s) do not have values in the right DataFrame.
merged_left = pd.merge(left=samples_table, right=geographic_info_table, how='left', left_on='Combined_Key', right_on='Combined_Key')
merged_left
```

```python
# Outer Join: returns the all pairwise combinations of rows from both DataFrames; i.e., the Cartesian product and the result DataFrame will use NaN where data is missing in one of the dataframes
merged_outer = pd.merge(left=samples_table, right=geographic_info_table, how='outer', left_on='Combined_Key', right_on='Combined_Key')
merged_outer
```

```python
# Give the function to the learners directly (no need for live coding)
# Tell them to check that identations are good

def create_batches(dataset:pd.DataFrame,
                   n_days_per_batch:int = 5)->list[pd.DataFrame]:
  """
  Split the COVID dataset into a list of subsets based on the 'Date' field.

  Args:
    dataset (pd.DataFrame): the COVID dataset.
    n_days_per_batch (int): the number of days in each batch.

  Returns:
    list[pd.DataFrame]: list of subsets of the COVID dataset.

  """
  # Recover all the days reported in the COVID dataset
  days = list(dataset['Date'].value_counts().keys())

  # Create a list where each element is another list of n_days_per_batch elements, whose elements are the days in the batch
  # Note: the last batch might have less elements than the others
  days_per_batch = [days[i:i+ n_days_per_batch] for i in range(0,len(days), n_days_per_batch)]

  batches = []
  for batch_idx in range(len(days_per_batch)):
    # Each batch is created by locating the rows where the 'Date' column matches one of the days of the batch
    batch = dataset[dataset['Date'].isin(days_per_batch[batch_idx])].copy()
    batches.append(batch)

  return batches

batches = create_batches(dataset)
batches[1] # Shows the data in the second batch (index 1)
```

```python
# matplotlib alone
fig, axs = plt.subplots(2, 2, figsize=(10, 10)) #start with this to show 4 empty windows

x = [i for i in range(10)] # then create these
y_1 = [i**2 for i in range(10)]
y_2 = [i**3 for i in range(10)]
y_3 = [i**4 for i in range(10)]
y_4 = [i**5 for i in range(10)]

axs[0,0].plot(x,y_1) # start without the labels
axs[0,0].set_title('y = x^2') # then add the labels
axs[0,1].plot(x,y_2)
axs[0,1].set_title('y = x^3')
axs[1,0].plot(x,y_3)
axs[1,0].set_title('y = x^4')
axs[1,1].plot(x,y_4)
axs[1,1].set_title('y = x^5')

plt.show()
```

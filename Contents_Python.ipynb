````python
# FYI: classes

class TableWithLists:
  """
  A simple table implemented with lists.

  Atributes:
    table(list): a list containing as many elements as the table has rows, which each element a list of as many elements the table has columns.

  """
  def __init__(self,
               n_columns: int,
               n_rows: int,
               default_value: int | float | str = 0) -> None:
    """
    Initialise the table.

    Args:
      n_columns (int): number of columns in the table.
      n_rows (int): number of rows in the table.
      default_value (int | float | str): default value for the elements of the table.

    Return:
      None.
    """
    self.table = []
    for i in range(n_rows):
      row = []
      for j in range(n_columns):
        row.append(default_value)
      self.table.append(row)

  def __str__(self)->str:
    """
    Special methods invoked when the object is called in a print().

    Args:
      None.

    Return:
      table_to_print (str): a string containing the elements of the table, organised by rows and columns.
    """
    table_to_print = ""
    for row in self.table:
      table_to_print += str(row) + "\n"
    return table_to_print

  def print_message(self,message:str)->None:
    print(message)


test_table = TableWithLists(n_rows = 3,n_columns=4)
print(test_table)

test_table.print_message('the message method worked!')
````

# Task 3 - Exercise: setup your first processing pipelining

````python
# Give the function to the learners directly (no need for live coding)
# Tell them to check that identations are good

import pandas as pd

def create_batches(dataset: pd.DataFrame,
                   n_days_per_batch: int = 5) -> list[pd.DataFrame]:
    """
    Split the dataset into batches of consecutive days based on the 'Date' column.

    Args:
        dataset (pd.DataFrame): Input dataset with a datetime64 'Date' column.
        n_days_per_batch (int): Number of days in each batch.

    Returns:
        list[pd.DataFrame]: List of subsets of the dataset, one per batch.
    """
    # Ensure Date is datetime and sorted
    dataset = dataset.copy()
    dataset = dataset.sort_values("Date")

    # Use Grouper to create time-based bins
    grouped = dataset.groupby(pd.Grouper(key="Date", freq=f"{n_days_per_batch}D"))

    # Collect non-empty groups
    return [group for _, group in grouped if not group.empty]


batches = create_batches(dataset)
batches[1] # Shows the data in the second batch (index 1)
````

````python
# DATA PREPARATION
def build_total_deaths_over_time(dataset: pd.DataFrame, n_days_per_batch: int = 5) -> pd.DataFrame:
    """
    Build a table of total COVID deaths per state, aggregated in batches of days.

    Args:
        dataset (pd.DataFrame): Input COVID dataset with 'Date', 'Province_State', 'Deaths'.
        n_days_per_batch (int): Number of days per batch.

    Returns:
        pd.DataFrame: Rows = batches, Columns = states, plus 'Batch_Day'.
    """
    # Ensure datetime and sorted
    dataset = dataset.sort_values("Date").copy()

    # Group by time batches and state
    grouped = (
        dataset
        .groupby([
            pd.Grouper(key="Date", freq=f"{n_days_per_batch}D"),
            "Province_State"
        ])["Deaths"]
        .sum()
        .reset_index()
    )

    # Pivot so states are columns
    pivoted = grouped.pivot_table(
        index="Date",
        columns="Province_State",
        values="Deaths",
        fill_value=0
    ).reset_index()

    # Add Batch_Day as number of days since monitoring start
    pivoted["Batch_Day"] = (pivoted["Date"] - pivoted["Date"].min()).dt.days

    return pivoted

n_days_per_batch=5
dataset_total_deaths_per_state = build_total_deaths_over_time(dataset, n_days_per_batch)
dataset_total_deaths_per_state
````

# Supplementary B - Machine Learning (time series prediction)

## B.1 Data preparation

Goal:


*   We want to predict the number of deaths in a state for a batch (a batch is an aggregate of concecutive days) based on the number of deaths in the previous batches.
*   This is called lag time series forcasting.



````python
# DATA PREPARATION
import numpy as np

def make_supervised_data(dataset_total_deaths_per_state: pd.DataFrame,
                         states_keys: list,
                         n_batches_to_predict_Y: int = 10):
    """
    Build supervised learning data (X, Y) for forecasting deaths.

    Args:
        dataset_total_deaths_per_state (pd.DataFrame): table with 1 row per batch, 1 column per state.
        states_keys (list): list of state names (columns to use).
        n_batches_to_predict_Y (int): number of lag batches to use as input.

    Returns:
        X (ndarray): shape (n_samples, n_batches_to_predict_Y, n_states), lag features.
        Y (ndarray): shape (n_samples, n_states), targets.
    """
    # Extract deaths as NumPy array (rows = batches, cols = states)
    data = dataset_total_deaths_per_state[states_keys].to_numpy()

    n_batches, n_states = data.shape
    n_samples = n_batches - n_batches_to_predict_Y

    # Build sliding windows
    X = np.array([data[i:i+n_batches_to_predict_Y] for i in range(n_samples)])
    Y = data[n_batches_to_predict_Y:]

    return X, Y

# Example usage
n_days_per_batch = 1
dataset_total_deaths_per_state = build_total_deaths_over_time(dataset, n_days_per_batch)[:-1]

X, Y = make_supervised_data(dataset_total_deaths_per_state, states_keys, n_batches_to_predict_Y=10)

print(X.shape)  # (n_samples, 10, n_states)
print(Y.shape)  # (n_samples, n_states)
````

## B.2 Create, train, and validate a simple model on all states

````python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

X, Y = make_supervised_data(dataset_total_deaths_per_state, states_keys, n_batches_to_predict_Y=10)

# Reshape: flatten lags × states into features
X = X.reshape(X.shape[0]*X.shape[2], -1)   # (samples*states, lags)
Y = Y.reshape(-1)  # (samples*states)

# Train-test split by dates
train_size = int(0.8 * len(X))
X_train_dates, X_test_dates = X[:train_size], X[train_size:]
Y_train_dates, Y_test_dates = Y[:train_size], Y[train_size:]


# Train model
model = LinearRegression()
model.fit(X_train_dates, Y_train_dates)

# Forecast
y_pred_train = model.predict(X_train_dates)
y_pred_test = model.predict(X_test_dates)

# Evaluation
train_mse = mean_squared_error(Y_train_dates, y_pred_train)
test_mse = mean_squared_error(Y_test_dates, y_pred_test)

train_rmse = np.sqrt(train_mse)
test_rmse = np.sqrt(test_mse)

train_r2 = r2_score(Y_train_dates, y_pred_train)
test_r2 = r2_score(Y_test_dates, y_pred_test)

print("Train MSE:", train_mse)
print("Test MSE:", test_mse)
print("Train RMSE:", train_rmse)
print("Test RMSE:", test_rmse)
print("Train R²:", train_r2)
print("Test R²:", test_r2)

# Scatter plot: predicted vs ground truth
plt.figure(figsize=(6,6))
plt.scatter(Y_train_dates, y_pred_train, alpha=0.2, label = "Train set")
plt.scatter(Y_test_dates, y_pred_test, alpha=0.7, label = "Test set")
plt.plot([Y_test_dates.min(), Y_test_dates.max()],
         [Y_test_dates.min(), Y_test_dates.max()],
         "r--", label="Perfect prediction")
plt.xlabel("Ground Truth (Actual Deaths)")
plt.ylabel("Predicted Deaths")
plt.title(f"Linear Regression Forecast")
plt.legend()
plt.grid(True)
plt.show()
````

## B.3 Impact of bias on predictions
